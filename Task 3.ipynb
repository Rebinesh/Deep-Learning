```python
import numpy as np
import tensorflow as tf

# Step 1: Create a small dataset using NumPy
np.random.seed(42)
input_data = np.random.randint(1, 100, size=(4, 5)).astype(np.float32)

print("Original Input:")
print(input_data)

# Step 2: Apply Batch Normalization
batch_norm = tf.keras.layers.BatchNormalization()
batch_output = batch_norm(input_data, training=True)

print("\nBatch Normalization Output:")
print(batch_output.numpy())

# Step 3: Apply Layer Normalization
layer_norm = tf.keras.layers.LayerNormalization()
layer_output = layer_norm(input_data)

print("\nLayer Normalization Output:")
print(layer_output.numpy())

# Step 4: Comparison
print("\nComparison:")
print("1. Batch Normalization normalizes data across the batch for each feature.")
print("2. Layer Normalization normalizes data across the features of each individual sample.")
print("3. Batch Normalization depends on the batch size, while Layer Normalization works independently of batch size.")
```
